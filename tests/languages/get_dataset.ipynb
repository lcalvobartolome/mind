```python
import pandas as pd
```

```python
subset = pd.read_parquet("/export/usuarios01/ivgomez/mind/t_data/inputs/subsets/subset312.parquet")
big = pd.read_parquet("/export/usuarios_ml4ds/lbartolome/Repos/umd/LinQAForge/data/source/corpus_rosie/passages/26_jan/df_1.parquet")
```

```python
subset = subset.merge(
    big[['doc_id', 'full_doc']],
    left_on='anchor_passage_id',
    right_on='doc_id',
    how='left'
).rename(columns={'full_doc': 'anchor_context'})
```

```python
subset = subset.drop(columns=['doc_id'])
```

```python
subset = subset.merge(
    big[['doc_id', 'full_doc']],
    left_on='comparison_passage_id',
    right_on='doc_id',
    how='left'
).rename(columns={'full_doc': 'comparison_context'})
```

```python
subset = subset.drop(columns=['doc_id'])
```

```python
subset.head()
```

```python
subset["anchor_context"].replace("", pd.NA, inplace=True)
subset["comparison_context"].replace("", pd.NA, inplace=True)
```

```python
subset["anchor_context"] = subset["anchor_context"].fillna(subset["anchor_passage"])
```

```python
subset["comparison_context"] = subset["comparison_context"].fillna(subset["comparison_passage"])
```

```python
subset.to_parquet("/export/usuarios01/ivgomez/mind/t_data/inputs/subsets/subset_231_contexto.parquet")
```
